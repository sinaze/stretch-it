```python
def grad(x, y):
    delta_x = np.array(x[1] - x[0])
    first = np.array((y[1] - y[0])) / delta_x
    middle = np.array(([y[i+2] - y[i] for i in range(0, len(y)-2)])) / (2*delta_x)
    last = np.array((y[-1] - y[-2])) / delta_x
    return np.concatenate(([first], middle, [last]))
```

```python
grad(np.array([1, 2, 3, 4, 5, 6]), y)
```
