```python
"""Analyze."""
import numpy as np
import pickle
import matplotlib.pyplot as plt
%matplotlib notebook


class ReducedSystem:
    """Reduced system only containing box size and averaged energies."""

    def __init__(self, mean_box, mean_en, std_en, err_en, dil, tot):
        """Initialize."""
        self.ener = mean_en
        self.std = std_en
        self.err = err_en
        self.box = mean_box
        self.dil = dil
        self.tot = tot
        self.phi = dil/tot
```

```python
dil = ['0', '5', '10', '15', '20', '35', '50', '75', '100', '125_e', '150_e']
files = []
for d in dil:
    fn = 'obj_' + d
    files.append(fn)
for f in files:
    print(f)
    print(eval(f).ener)
    print('\n')
```

```python
phase_line = []
last_idx = [0, 1, 3, 5, 7, 6, 18, 14, 17, 1, 4]
for f, i in zip(files, last_idx):
    phase_line.append(eval(f).box[i, 1])
phase_line = np.array(phase_line)
```

```python
dil_a = np.array([0, 5, 10, 15, 20, 30, 50, 75, 100, 125, 150])
linfit = np.polyfit(phase_line, dil_a/512*100, 1)
fit_f = np.poly1d(linfit)
X = np.linspace(24, 32)
plt.figure()
plt.plot(phase_line, dil_a/512*100, 'kx')
plt.plot(X, fit_f(X), 'k-')
plt.fill_between(X, fit_f(X), color=plt.cm.Pastel2(1))
maxy = plt.ylim()[1]
plt.fill_between(X, fit_f(X), maxy, color=plt.cm.Pastel2(0))
plt.text(25, 20, "floppy", fontsize=14)
plt.text(28, 5, "elastic", fontsize=14)
plt.title('Phase Diagram')
plt.ylabel('dilution fraction (%)')
plt.xlabel('$L_y$')
plt.show()
```
