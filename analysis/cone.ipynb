# Use new formula for cone model (one layer)

```python
def modulus_m(phi, young_func, area0):
    const = np.square(0.8/area0)
    res = 0.5 * young_func(phi) * const
    return res
```

# multi-layer formula

```python
dz = 1/100
```

```python
def quad_sum(z, dz):
    s = 0
    for n in range(0, int(z/dz+1)):
        s += np.square(z - n*dz)
    return np.square(s)
```

```python
vquad_sum = np.vectorize(quad_sum)
```

```python
Z = np.linspace(0, 1)
plt.figure()
plt.plot(Z, modulus_m(0, fit, 1)*vquad_sum(Z, dz), '-', label=r'$\phi = 0$ %')
plt.plot(Z, modulus_m(10, fit, 1)*vquad_sum(Z, dz), '-', label=r'$\phi = 10$ %')
plt.plot(Z, modulus_m(20, fit, 1)*vquad_sum(Z, dz), '-', label=r'$\phi = 20$ %')
plt.plot(Z, modulus_m(30, fit, 1)*vquad_sum(Z, dz), '-', label=r'$\phi = 30$ %')
plt.legend()
plt.xlabel('indentation $z$ ($\mu$m)')
plt.ylabel('$U(z)$')
plt.show()
```
