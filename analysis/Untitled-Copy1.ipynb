```python
"""Analyze."""
import numpy as np
import pickle
import matplotlib.pyplot as plt
%matplotlib notebook


class ReducedSystem:
    """Reduced system only containing box size and averaged energies."""

    def __init__(self, mean_box, mean_en, std_en, err_en, dil, tot):
        """Initialize."""
        self.ener = mean_en
        self.std = std_en
        self.err = err_en
        self.box = mean_box
        self.dil = dil
        self.tot = tot
        self.phi = dil/tot
```

```python
plt.figure()
plt.plot(X1, Y1, '.-')
plt.plot(X2-9, Y2, '.-')
plt.plot(X3-15, Y3, '.-')
plt.show()
```

```python
p, V = np.polyfit(X1, Y1, 2, cov=True)
print("curvature: {} +/- {}".format(2*p[0], 2*np.sqrt(V[0][0])))
fit = np.poly1d(p)
X = np.linspace(15, 18)
plt.figure()
plt.plot(X1, Y1, '.-')
plt.plot(X, fit(X))
plt.plot()
plt.show()
```
