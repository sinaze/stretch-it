```python
"""Analyze."""
import numpy as np
import pickle
import matplotlib.pyplot as plt
%matplotlib notebook


class ReducedSystem:
    """Reduced system only containing box size and averaged energies."""

    def __init__(self, mean_box, mean_en, std_en, err_en, dil, tot):
        """Initialize."""
        self.ener = mean_en
        self.std = std_en
        self.err = err_en
        self.box = mean_box
        self.dil = dil
        self.tot = tot
        self.phi = dil/tot
```
